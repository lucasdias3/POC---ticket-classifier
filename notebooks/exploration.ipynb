# Ticket Classification System Exploration

This notebook is used for exploratory data analysis (EDA) on the ticket dataset from the `knowledge_base.json`. The goal is to gain insights into the data that will inform model selection and feature engineering for the ticket classification system.

```python
import pandas as pd
import json
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
with open('../data/knowledge_base.json', 'r', encoding='utf-8') as f:
    data = json.load(f)

# Convert to DataFrame
df = pd.DataFrame(data)

# Display the first few rows of the dataset
df.head()
```

```python
# Check the shape of the dataset
df.shape
```

```python
# Check for missing values
df.isnull().sum()
```

```python
# Visualize the distribution of ticket categories
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='category', order=df['category'].value_counts().index)
plt.title('Distribution of Ticket Categories')
plt.xticks(rotation=45)
plt.show()
```

```python
# Visualize the distribution of ticket priorities
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='priority', order=df['priority'].value_counts().index)
plt.title('Distribution of Ticket Priorities')
plt.xticks(rotation=45)
plt.show()
```

```python
# Analyze the relationship between customer segment and ticket category
plt.figure(figsize=(12, 8))
sns.countplot(data=df, x='customer_segment', hue='category')
plt.title('Customer Segment vs Ticket Category')
plt.xticks(rotation=45)
plt.legend(title='Category')
plt.show()
```

## Insights
1. The dataset contains a variety of ticket categories, with some categories being more prevalent than others.
2. Ticket priorities are distributed across different levels, indicating varying urgency among tickets.
3. Customer segments show distinct patterns in ticket categories, which may inform feature engineering for the classification model.
